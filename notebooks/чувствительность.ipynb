```python
long_data['Разница_цен'] = (long_data['Наша_цена'] - long_data['Цена_конкурентов'])
long_data['Изменение_разницы'] = long_data['Разница_цен'] - long_data['Разница_цен'].shift(1)
long_data['Изменение_объема_л'] = long_data['Объем_л'] - long_data['Объем_л'].shift(1)
previous = long_data['Объем_л'].shift(1).where(long_data['Объем_л'].shift(1) > 0)
long_data['Изменение_объема_pct'] = (long_data['Изменение_объема_л'] / previous * 100)
long_data.head(2)
```

```python
events = long_data.dropna(subset=['Изменение_разницы', 'Изменение_объема_л']).copy() # у каждого клиента удаляю первый месяц # стоп хуйню сделала 
events['Ухудшение_цены'] = events['Изменение_разницы'] > 0
events['Сокращение_при_ухудшении'] = (events['Ухудшение_цены']& (events['Изменение_объема_л'] < 0))
events['Без_ухудшения'] = events['Изменение_разницы'] <= 0
events['Сокращение_без_ухудшения'] = (events['Без_ухудшения']& (events['Изменение_объема_л'] < 0))
events
```
